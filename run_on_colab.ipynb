# Run Streamlit App on Google Colab
This notebook helps you run your Streamlit app (`streamlit_app.py`) on Google Colab using Cloudflare Tunnels (which avoids the localtunnel loading issues).

### Step 1: Clone the Private Repository
To clone your private repository, you need a GitHub Personal Access Token (PAT). Create one at https://github.com/settings/tokens (needs `repo` scope).
Then, use Colab's built-in Secrets manager (the key icon on the left sidebar) to add a new secret named `GITHUB_TOKEN` and paste your token as the value.

```python
import os
try:
    from google.colab import userdata
    github_token = userdata.get('GITHUB_TOKEN')
except:
    github_token = input("Enter your GitHub Personal Access Token (PAT): ")

GITHUB_USER = 'usman2003'
REPO_NAME = 'SmartStoreVisionAI'

repo_url = f'https://{github_token}@github.com/{GITHUB_USER}/{REPO_NAME}.git'

!git clone {repo_url}
%cd {REPO_NAME}
```

### Step 3: Run Streamlit & Cloudflare Tunnel
This will start the Streamlit app in the background and use Cloudflare to expose it. Cloudflare doesn't have an interstitial page, so all Streamlit components will load perfectly.
**Look for the link ending in `.trycloudflare.com` in the output below!**

```python
# Download Cloudflare tunnel binary
!wget -q -nc https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64
!chmod +x cloudflared-linux-amd64

# Start Streamlit in the background
import subprocess
subprocess.Popen(["streamlit", "run", "streamlit_app.py", "--server.port", "8501", "--server.headless", "true"])

# Start Cloudflare tunnel
import time
time.sleep(3) # Give Streamlit a few seconds to start
!./cloudflared-linux-amd64 tunnel --url http://127.0.0.1:8501
```
